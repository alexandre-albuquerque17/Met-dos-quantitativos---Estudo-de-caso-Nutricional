Pergunta problema:
* Existem relação significativa entre o estado nutricional de mulheres e homens adultos entre os anos de 2008 até 2025 em Alagoas?
---

Objetivo geral:
* Verificar se no período analisado (2008 - 2025) houve correlação nos estados nutricionais entre os sexos masculinos e femininos no estado de Alagoas.
----

Objetivos específicos:

* Apresentar a estatística descritiva dos dados e aplicação dos testes (Mann Whithey U e Spearmarn) nas variavéis baixo peso, adequado, sobrepeso, obesidade de grau I, II e III no período entre 2008 e 2025 para homens e mulheres no Estado através da manipulação de tabelas e gráficos.

---

Hipóteses:

H0: Através do teste Mann-Wwhitney: Há diferenças entre o estado nutricional de mulheres e homens (Baixo Peso, Peso Adequado, Sobrepeso, Obsedidade Grau I, Obesidade Grau II e Obesidade Grau III) em Alagoas.

H1: Através do teste Mann-Wwhitney: Não há diferenças entre o estado nutricional de mulheres e homens (Baixo Peso, Peso Adequado, Sobrepeso, Obsedidade Grau I, Obesidade Grau II e Obesidade Grau III) em Alagoas.

----

H0: Através do teste Spearman: Há relação entre o estado nutricional de homens e mulheres (Baixo Peso, Peso Adequado, Sobrepeso, Obsedidade Grau I, Obesidade Grau II e Obesidade Grau III) em Alagoas.

H1: Através do teste Spearman: Não há relação entre o estado nutricional de homens e mulheres (Baixo Peso, Peso Adequado, Sobrepeso, Obsedidade Grau I, Obesidade Grau II e Obesidade Grau III)  em Alagoas.

Teste Mann-Whitney U (dados não parâmetricos):


- Verificando se há diferença entre o estado nutricional entre mulheres e homens envolvendo todas variavéis nutricionais (subnutrição, sobrepeso e obesidade).


Dado que a ordem não importa iniciamos o teste referente a mulheres.

- Em suma, no teste Mann-Whitney U, a ordem das amostras não altera o p-valor, resultado que verifica diferenças significativas entre os grupos, muda-se apenas o valor da estatística U, caso inverta, o valor de U será referente aos homens.

---
Explicação do código:

- variável - baixo_peso_estatistica, baixo_peso_p_valor

São as variáveis que vão receber os resultados do teste

- = stats.mannwhitneyu() - teste
   
   
Função que aplica o teste estatístico Mann-Whitney U

- df_filtrado_col_sel['Quantidade de mulheres com baixo peso'] - pega tabela e coluna específica
   
   
Primeira amostra: valores da coluna de mulheres

- df_filtrado_col_sel['Quantidade de homens com baixo peso'] - pega tabela e coluna específica
   
Segunda amostra: valores da coluna de homens.

- if - se condicinal
* p_valor de baixo peso < 0,05 representa importância estatística, caso contrário > 0,05 sem importância.

- Mediana é uma tendência central (entre média e moda) mostrando qual grupo tem a tendência a ter valores maiores.

Exemplo: mediana feminina > masculina → tendência maior ao baixo peso em mulheres.

- A magnitude é dado pelo tamanho da diferença entre as medianas dos sexos dentre o aspecto nutricional vigente de cada grupo.


- Intervalo interquantil (IQR) mede a dispersão dos dados. Sendo assim, mostra se os valores estão concentrados ou espalhados.



Esta aplicação é usada para as outras variavéis importantes envolvendo a nutriçãos dos agentes.

```python
#Teste Mann-Whitney para o estado nutrional considerado baixo peso:

baixo_peso_estatistica, baixo_peso_p_valor = stats.mannwhitneyu(df_filtrado_col_sel['Quantidade de mulheres com baixo peso'], df_filtrado_col_sel['Quantidade de homens com baixo peso'])

#Realizando o condicional para o p_valor:

if baixo_peso_p_valor < 0.05:
  resultado = 'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que há uma diferença estatisticamente significante entre os estados nutrionais de homens e mulheres com baixo peso'
  hipotese = 'Logo aceita-se a H0, pois há diferenças estatisticamente comprovadas entre o baixo peso entre homens e mulheres'
else:
   resultado = 'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que não há uma diferença estatisticamente significante entre os estados nutricionais de homens e mulheres com de baixo peso, podendo ser coincidência as diferenças registradas'
   hipotese = 'Logo rejeita-se a H0, pois não há diferenças estatisticamente comprovadas entre o baixo peso entre homens e mulheres'

#Realizando as medianas:

mediana_bx_peso_fem = df_filtrado_col_sel['Quantidade de mulheres com baixo peso'].median()
mediana_bx_peso_mas = df_filtrado_col_sel['Quantidade de homens com baixo peso'].median()

#Diferença entre as medianas:

diferença_medianas = mediana_bx_peso_fem - mediana_bx_peso_mas

#Realizando o intervalo de interquantil:

interquantil_bx_peso_fem_25, interquantil_bx_peso_fem_75 = df_filtrado_col_sel['Quantidade de mulheres com baixo peso'].quantile([0.25, 0.75])
interquantil_bx_peso_mas_25, interquantil_bx_peso_mas_75 = df_filtrado_col_sel['Quantidade de homens com baixo peso'].quantile([0.25, 0.75])

#Realizando a diferença entre os o intervalo interquantil para saber o tamanho da dispersão dos dados:

tamanho_dispersão_bx_peso_fem = interquantil_bx_peso_fem_75 - interquantil_bx_peso_fem_25
tamanho_dispersão_bx_peso_mas = interquantil_bx_peso_mas_75 - interquantil_bx_peso_mas_25

#Realizando o condicional entre a mediana de mulheres e homens:

if mediana_bx_peso_fem > mediana_bx_peso_mas:
  mediana = f'A condição de baixo peso feminino é maior em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, as mulheres tendem a ter maior tendencia a ter baixo peso.'
else:
  mediana = f'A condição de baixo peso feminino é menor em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, os homens tendem a ter maior tendencia a ter baixo peso.'

#Realizando o condicional entre os intervalos interquantis:

if tamanho_dispersão_bx_peso_fem > tamanho_dispersão_bx_peso_mas:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de baixo peso das mulheres é maior do que a de homens.'
else:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de baixo peso das mulheres é menor do que a de homens.'

print(f'Resultado das medianas: Mulheres {mediana_bx_peso_fem} e homens {mediana_bx_peso_mas}.\n'
      f'{mediana}\n'
      f'Resultado dos intervalos interquantis: Mulheres {interquantil_bx_peso_fem_25, interquantil_bx_peso_fem_75} e homens {interquantil_bx_peso_mas_25, interquantil_bx_peso_mas_75}.\n'
      f'{interquantil}\n'
      f'Resultado teste Mann-Whitney: {baixo_peso_estatistica:.4f} | P-valor {baixo_peso_p_valor:.4f}\n'
      f'{resultado}\n'
      f'{hipotese}\n'
)

#____________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________
#Teste Mann-Whitney para o estado nutrional considerado adequado:

peso_adequado_estatistica, peso_adequado_p_valor = stats.mannwhitneyu(df_filtrado_col_sel['Quantidade de mulheres com peso adequado'], df_filtrado_col_sel['Quantidade de homens com peso adequado'])

#Realizando o condicional para o p_valor:

if peso_adequado_p_valor < 0.05:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que há uma diferença entre os estados nutrionais de homens e mulheres com peso adequado'
  hipotese = 'Logo aceita-se a H0, pois há diferenças estatisticamente comprovadas entre o peso adequado entre homens e mulheres'
else:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que não há uma diferença estatistica entre os estados nutricionais de homens e mulheres com peso considerado adequado, podendo ser coincidência as diferenças registradas'
  hipotese = 'Logo rejeita-se a H0, pois não há diferenças estatisticamente comprovadas entre o peso adequado entre homens e mulheres'
#Realizando as medianas:

mediana_peso_adequado_fem = df_filtrado_col_sel['Quantidade de mulheres com peso adequado'].median()
mediana_peso_adequado_mas = df_filtrado_col_sel['Quantidade de homens com peso adequado'].median()

#Diferença entre as medianas:

diferença_medianas = mediana_peso_adequado_fem - mediana_peso_adequado_mas

#Realizando o intervalo de interquantil:

interquantil_peso_adequado_fem_25, interquantil_peso_adequado_fem_75 = df_filtrado_col_sel['Quantidade de mulheres com peso adequado'].quantile([0.25, 0.75])
interquantil_peso_adequado_mas_25, interquantil_peso_adequado_mas_75 = df_filtrado_col_sel['Quantidade de homens com peso adequado'].quantile([0.25, 0.75])

#Realizando a diferença entre os o intervalo interquantil para saber o tamanho da dispersão dos dados:

tamanho_dispersão_peso_adequado_fem = interquantil_peso_adequado_fem_75 - interquantil_peso_adequado_fem_25
tamanho_dispersão_peso_adequado_mas = interquantil_peso_adequado_mas_75 - interquantil_peso_adequado_mas_25

#Realizando o condicional entre a mediana de mulheres e homens:

if mediana_peso_adequado_fem > mediana_peso_adequado_mas:
  mediana = f'A condição de peso adequado feminino é maior em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, as mulheres tendem a ter maior tendencia a ter o peso adequado.'
else:
  mediana = f'A condição de peso adequado feminino é menor em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, os homens tendem a ter maior tendencia a ter o peso adequado.'

#Realizando o condicional entre os intervalos interquantis:

if tamanho_dispersão_peso_adequado_fem > tamanho_dispersão_peso_adequado_mas:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de peso adequado das mulheres é maior do que a de homens.'
else:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de peso adequado das mulheres é menor do que a de homens.'

print(f'Resultado das medianas: Mulheres {mediana_peso_adequado_fem} e homens {mediana_peso_adequado_mas}.\n'
      f'{mediana}\n'
      f'Resultado dos intervalos interquantis: Mulheres {interquantil_peso_adequado_fem_25, interquantil_peso_adequado_fem_75} e homens {interquantil_peso_adequado_mas_25, interquantil_peso_adequado_mas_75}.\n'
      f'{interquantil}\n'
      f'Resultado teste Mann-Whitney: {peso_adequado_estatistica:.4f} | P-valor {peso_adequado_p_valor:.4f}\n'
      f'{resultado}\n'
      f'{hipotese}\n')

#___________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________
#Teste Mann-Whitney para o estado nutrional considerado sobrepeso:

sobrepeso_estatistica, sobrepeso_p_valor = stats.mannwhitneyu(df_filtrado_col_sel['Quantidade de mulheres com sobrepeso'], df_filtrado_col_sel['Quantidade de homens com sobrepeso'])

#Realizando o condicional para o p_valor:

if sobrepeso_p_valor < 0.05:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que há uma diferença entre os estados nutrionais de homens e mulheres com sobrepeso'
  hipotese = 'Logo aceita-se a H0, pois há diferenças estatisticamente comprovadas entre o sobrepeso entre homens e mulheres'
else:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que não há uma diferença estatistica entre os estados nutricionais de homens e mulheres com sobrepeso, podendo ser coincidência as diferenças registradas'
  hipotese = 'Logo rejeita-se a H0, pois não há diferenças estatisticamente comprovadas entre o sobrepeso entre homens e mulheres'

#Realizando as medianas:

mediana_sobrepeso_fem = df_filtrado_col_sel['Quantidade de mulheres com sobrepeso'].median()
mediana_sobrepeso_mas = df_filtrado_col_sel['Quantidade de homens com sobrepeso'].median()

#Diferença entre as medianas:

diferença_medianas = mediana_sobrepeso_fem - mediana_sobrepeso_mas

#Realizando o intervalo de interquantil:

interquantil_sobrepeso_fem_25, interquantil_sobrepeso_fem_75 = df_filtrado_col_sel['Quantidade de mulheres com sobrepeso'].quantile([0.25, 0.75])
interquantil_sobrepeso_mas_25, interquantil_sobrepeso_mas_75 = df_filtrado_col_sel['Quantidade de homens com sobrepeso'].quantile([0.25, 0.75])

#Realizando a diferença entre os o intervalo interquantil para saber o tamanho da dispersão dos dados:

tamanho_dispersão_sobrepeso_fem = interquantil_sobrepeso_fem_75 - interquantil_sobrepeso_fem_25
tamanho_dispersão_sobrepeso_mas = interquantil_sobrepeso_mas_75 - interquantil_sobrepeso_mas_25

#Realizando o condicional entre a mediana de mulheres e homens:

if mediana_sobrepeso_fem > mediana_sobrepeso_mas:
  mediana = f'A condição de sobrepeso feminino é maior em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, as mulheres tendem a ter maior tendencia a ter o sobrepeso.'
else:
  mediana = f'A condição de sobrepeso feminino é menor em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, os homens tendem a ter maior tendencia a ter o sobrepeso.'

#Realizando o condicional entre os intervalos interquantis:

if tamanho_dispersão_sobrepeso_fem > tamanho_dispersão_sobrepeso_mas:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de sobrepeso das mulheres é maior do que a de homens.'
else:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de sobrepeso das mulheres é menor do que a de homens.'

print(f'Resultado das medianas: Mulheres {mediana_sobrepeso_fem} e homens {mediana_sobrepeso_mas}.\n'
      f'{mediana}\n'
      f'Resultado dos intervalos interquantis: Mulheres {interquantil_sobrepeso_fem_25, interquantil_sobrepeso_fem_75} e homens {interquantil_sobrepeso_mas_25, interquantil_sobrepeso_mas_75}.\n'
      f'{interquantil}\n'
      f'Resultado teste Mann-Whitney: {sobrepeso_estatistica:.4f} | P-valor {sobrepeso_p_valor:.4f}\n'
      f'{resultado}\n'
      f'{hipotese}\n')

#___________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________
#Teste Mann-Whitney para o estado nutrional considerado obesidade grau I:


obesidade_grau_I_estatistica, obesidade_grau_I_p_valor = stats.mannwhitneyu(df_filtrado_col_sel['Quantidade de mulheres com obesidade grau I'], df_filtrado_col_sel['Quantidade de homens com obesidade grau I'])

#Realizando o condicional para o p_valor:

if obesidade_grau_I_p_valor < 0.05:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que há uma diferença entre os estados nutrionais de homens e mulheres com obesidade grau I'
  hipotese = 'Logo aceita-se a H0, pois há diferenças estatisticamente comprovadas entre o obesidade grau I entre homens e mulheres'
else:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que não há uma diferença estatistica entre os estados nutricionais de homens e mulheres com obesidade grau I, podendo ser coincidência as diferenças registradas'
  hipotese = 'Logo rejeita-se a H0, pois não há diferenças estatisticamente comprovadas entre o obesidade grau I entre homens e mulheres'
#Realizando as medianas:

mediana_obesidade_grau_I_fem = df_filtrado_col_sel['Quantidade de mulheres com obesidade grau I'].median()
mediana_obesidade_grau_I_mas = df_filtrado_col_sel['Quantidade de homens com obesidade grau I'].median()

#Diferença entre as medianas:

diferença_medianas = mediana_obesidade_grau_I_fem - mediana_obesidade_grau_I_mas

#Realizando o intervalo de interquantil:

interquantil_obesidade_grau_I_fem_25, interquantil_obesidade_grau_I_fem_75 = df_filtrado_col_sel['Quantidade de mulheres com obesidade grau I'].quantile([0.25, 0.75])
interquantil_obesidade_grau_I_mas_25, interquantil_obesidade_grau_I_mas_75 = df_filtrado_col_sel['Quantidade de homens com obesidade grau I'].quantile([0.25, 0.75])

#Realizando a diferença entre os o intervalo interquantil para saber o tamanho da dispersão dos dados:

tamanho_dispersão_obesidade_grau_I_fem = interquantil_obesidade_grau_I_fem_75 - interquantil_obesidade_grau_I_fem_25
tamanho_dispersão_obesidade_grau_I_mas = interquantil_obesidade_grau_I_mas_75 - interquantil_obesidade_grau_I_mas_25

#Realizando o condicional entre a mediana de mulheres e homens:

if mediana_obesidade_grau_I_fem > mediana_obesidade_grau_I_mas:
  mediana = f'A condição de obesidade grau I feminino é maior em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, as mulheres tendem a ter maior tendencia a ter o obesidade grau I.'
else:
  mediana = f'A condição de obesidade grau I feminino é menor em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, os homens tendem a ter maior tendencia a ter o obesidade grau I.'

#Realizando o condicional entre os intervalos interquantis:

if tamanho_dispersão_obesidade_grau_I_fem > tamanho_dispersão_obesidade_grau_I_mas:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de obesidade grau I das mulheres é maior do que a de homens.'
else:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de obesidade grau I das mulheres é menor do que a de homens.'

print(f'Resultado das medianas: Mulheres {mediana_obesidade_grau_I_fem} e homens {mediana_obesidade_grau_I_mas}.\n'
      f'{mediana}\n'
      f'Resultado dos intervalos interquantis: Mulheres {interquantil_obesidade_grau_I_fem_25, interquantil_obesidade_grau_I_fem_75} e homens {interquantil_obesidade_grau_I_mas_25, interquantil_obesidade_grau_I_mas_75}.\n'
      f'{interquantil}\n'
      f'Resultado: {obesidade_grau_I_estatistica:.4f} | P-valor {obesidade_grau_I_p_valor:.4f}\n'
      f'{resultado}\n'
      f'{hipotese}\n')

#___________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________
#Teste Mann-Whitney para o estado nutrional considerado obesidade grau II:


obesidade_grau_II_estatistica, obesidade_grau_II_p_valor = stats.mannwhitneyu(df_filtrado_col_sel['Quantidade de mulheres com obesidade grau II'], df_filtrado_col_sel['Quantidade de homens com obesidade grau II'])

#Realizando o condicional para o p_valor:

if obesidade_grau_II_p_valor < 0.05:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que há uma diferença entre os estados nutrionais de homens e mulheres com obesidade grau II'
  hipotese = 'Logo aceita-se a H0, pois há diferenças estatisticamente comprovadas entre o obesidade grau II entre homens e mulheres'
else:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que não há uma diferença estatistica entre os estados nutricionais de homens e mulheres com obesidade grau II, podendo ser coincidência as diferenças registradas'
  hipotese = 'Logo rejeita-se a H0, pois não há diferenças estatisticamente comprovadas entre o obesidade grau II entre homens e mulheres'

#Realizando as medianas:

mediana_obesidade_grau_II_fem = df_filtrado_col_sel['Quantidade de mulheres com obesidade grau II'].median()
mediana_obesidade_grau_II_mas = df_filtrado_col_sel['Quantidade de homens com obesidade grau II'].median()

#Diferença entre as medianas:

diferença_medianas = mediana_obesidade_grau_II_fem - mediana_obesidade_grau_II_mas

#Realizando o intervalo de interquantil:

interquantil_obesidade_grau_II_fem_25, interquantil_obesidade_grau_II_fem_75 = df_filtrado_col_sel['Quantidade de mulheres com obesidade grau II'].quantile([0.25, 0.75])
interquantil_obesidade_grau_II_mas_25, interquantil_obesidade_grau_II_mas_75 = df_filtrado_col_sel['Quantidade de homens com obesidade grau II'].quantile([0.25, 0.75])

#Realizando a diferença entre os o intervalo interquantil para saber o tamanho da dispersão dos dados:

tamanho_dispersão_obesidade_grau_II_fem = interquantil_obesidade_grau_II_fem_75 - interquantil_obesidade_grau_II_fem_25
tamanho_dispersão_obesidade_grau_II_mas = interquantil_obesidade_grau_II_mas_75 - interquantil_obesidade_grau_II_mas_25

#Realizando o condicional entre a mediana de mulheres e homens:

if mediana_obesidade_grau_II_fem > mediana_obesidade_grau_II_mas:
  mediana = f'A condição de obesidade grau II feminino é maior em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, as mulheres tendem a ter maior tendencia a ter o obesidade grau II.'
else:
  mediana = f'A condição de obesidade grau II feminino é menor em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, os homens tendem a ter maior tendencia a ter o obesidade grau II.'

#Realizando o condicional entre os intervalos interquantis:

if tamanho_dispersão_obesidade_grau_II_fem > tamanho_dispersão_obesidade_grau_II_mas:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de obesidade grau II das mulheres é maior do que a de homens.'
else:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de obesidade grau II das mulheres é menor do que a de homens.'

print(f'Resultado das medianas: Mulheres {mediana_obesidade_grau_II_fem} e homens {mediana_obesidade_grau_II_mas}.\n'
      f'{mediana}\n'
      f'Resultado dos intervalos interquantis: Mulheres {interquantil_obesidade_grau_II_fem_25, interquantil_obesidade_grau_II_fem_75} e homens {interquantil_obesidade_grau_II_mas_25, interquantil_obesidade_grau_II_mas_75}.\n'
      f'{interquantil}\n'
      f'Resultado teste Mann-Whitney: {obesidade_grau_II_estatistica:.4f} | P-valor {obesidade_grau_II_p_valor:.4f}\n'
      f'{resultado}\n'
      f'{hipotese}\n')

#___________________________________________________________________________________________________________________________________________________________________________________________________________________________________________________
#Teste Mann-Whitney para o estado nutrional considerado obesidade grau III:


obesidade_grau_III_estatistica, obesidade_grau_III_p_valor = stats.mannwhitneyu(df_filtrado_col_sel['Quantidade de mulheres com obesidade grau III'], df_filtrado_col_sel['Quantidade de homens com obesidade grau III'])

#Realizando o condicional para o p_valor:

if obesidade_grau_III_p_valor < 0.05:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que há uma diferença entre os estados nutrionais de homens e mulheres com obesidade grau III'
  hipotese = 'Logo aceita-se a H0, pois há diferenças estatisticamente comprovadas entre o obesidade grau III entre homens e mulheres'
else:
  resultado = f'Por fim o resultado do p_valor obtido pelo teste de Mann-Whitney, demonstra que não há uma diferença estatistica entre os estados nutricionais de homens e mulheres com obesidade grau III, podendo ser coincidência as diferenças registradas'
  hipotese = 'Logo rejeita-se a H0, pois não há diferenças estatisticamente comprovadas entre o obesidade grau III entre homens e mulheres'

#Realizando as medianas:

mediana_obesidade_grau_III_fem = df_filtrado_col_sel['Quantidade de mulheres com obesidade grau III'].median()
mediana_obesidade_grau_III_mas = df_filtrado_col_sel['Quantidade de homens com obesidade grau III'].median()

#Diferença entre as medianas:

diferença_medianas = mediana_obesidade_grau_III_fem - mediana_obesidade_grau_III_mas

#Realizando o intervalo de interquantil:

interquantil_obesidade_grau_III_fem_25, interquantil_obesidade_grau_III_fem_75 = df_filtrado_col_sel['Quantidade de mulheres com obesidade grau III'].quantile([0.25, 0.75])
interquantil_obesidade_grau_III_mas_25, interquantil_obesidade_grau_III_mas_75 = df_filtrado_col_sel['Quantidade de homens com obesidade grau III'].quantile([0.25, 0.75])

#Realizando a diferença entre os o intervalo interquantil para saber o tamanho da dispersão dos dados:

tamanho_dispersão_obesidade_grau_III_fem = interquantil_obesidade_grau_III_fem_75 - interquantil_obesidade_grau_III_fem_25
tamanho_dispersão_obesidade_grau_III_mas = interquantil_obesidade_grau_III_mas_75 - interquantil_obesidade_grau_III_mas_25

#Realizando o condicional entre a mediana de mulheres e homens:

if mediana_obesidade_grau_III_fem > mediana_obesidade_grau_III_mas:
  mediana = f'A condição de obesidade grau III feminino é maior em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, as mulheres tendem a ter maior tendencia a ter o obesidade grau III.'
else:
  mediana = f'A condição de obesidade grau III feminino é menor em magnetude bruta do que o masculino, dado que a diferença absoluta das medianas é de {diferença_medianas}, ou seja, os homens tendem a ter maior tendencia a ter o obesidade grau III.'

#Realizando o condicional entre os intervalos interquantis:

if tamanho_dispersão_obesidade_grau_III_fem > interquantil_obesidade_grau_III_mas_25:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de obesidade grau III das mulheres é maior do que a de homens.'
else:
  interquantil = 'Ademais, o intervalo interquantil demonstra que a dispersão da condição de obesidade grau III das mulheres é menor do que a de homens.'

print(f'Resultado das medianas: Mulheres {mediana_obesidade_grau_III_fem} e homens {mediana_obesidade_grau_III_mas}.\n'
      f'{mediana}\n'
      f'Resultado dos intervalos interquantis: Mulheres {interquantil_obesidade_grau_III_fem_25, interquantil_obesidade_grau_III_fem_75} e homens {interquantil_obesidade_grau_III_mas_25, interquantil_obesidade_grau_III_mas_75}.\n'
      f'{interquantil}\n'
      f'Resultado teste Mann-Whitney: {obesidade_grau_III_estatistica:.4f} | P-valor {obesidade_grau_III_p_valor:.4f}\n'
      f'{resultado}\n'
      f'{hipotese}\n')
```

Teste de correlação de Spearman (dados não-paramétricos)

- É usado para compreender associação entre duas variáveis e avalia de acordo com a transformação em posições se quando uma variável aumenta, a outra também tende a aumentar ou diminuir.

---

Sobre o código:

- variável - bx_peso_coeficiente, bx_peso_p_valor - Servem como receptáculos dos valores calculados do teste

- stats.spermanr() - É um submódulo que realiza o teste e mede a relação (correlação) entre duas variáveis

Nele há argumentos

O primeiro argumento - seleciona o DataFrame[junto com a coluna], seguindo a padronização do teste anterior de Mann, optamos que a primeira amostra detenha a coluna de mulheres

- df_filtrado_col_sel['Quantidade de mulheres com baixo peso'] - pega tabela e coluna específica


- df_filtrado_col_sel['Quantidade de homens com baixo peso'] Segunda amostra: valores da coluna de homens | pega tabela e coluna específica


- print(f"Coeficiente: {bx_peso_coeficiente} | P-valor: {bx_peso_p_valor}") - Mostra o coeficiente de correlação e o p-valor

- if (se) - condicional - bx_peso_p_valor < 0.05 - Se o p-valor for menor que 0.05 → existe relação significativa entre as variáveis
